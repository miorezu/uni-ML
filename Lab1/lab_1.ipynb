```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("google_review_ratings.csv")
df.head()
```

```python
print(df.info())
```

```python
missing_values = df.isnull().sum()
print(missing_values)
print(df.columns)
#see unnamed column with all missing values. I guess this is because of the extra coma at the end of each line.
```

```python
#Delete this extra column and check again
df = df.drop(columns=df.columns[-1])
missing_values = df.isnull().sum()
print(missing_values)
print(df.columns)
```

```python
df = df.fillna(df.mean(numeric_only=True))
print("\nMissing values after filling only numeric:")
print(df.isnull().sum())

df = df.drop_duplicates()
df.drop("User", axis=1, inplace=True) #Delete "User" column

#Clear all lines in the data from unnecessary whitespace
df = df.apply(lambda col: col.str.replace('\t', '').str.strip() if col.dtype == "object" else col)
df = df.astype(float)
print(df.dtypes)
df.to_csv("cleaned_google_review_ratings.csv", index=False)
```

Cultural venues (museums, theatres) tend to have moderately high ratings with less polarization.
The correlation patterns suggest that people who rate certain venue types highly tend to rate similar venue types highly as well (e.g.,parks and theatres, restaurants and pubs).
Most categories show a "perfect score effect" with spikes at the maximum rating of 5, suggesting many reviewers give perfect scores regardless of category.

## Task 3: Feature Engineering
**Objective:** Create new features to improve the dataset's predictive power.

**Instructions:**

```python
cd = pd.read_csv("cleaned_google_review_ratings.csv")
cd.head()
```

```python
cd['Leisure_Rating'] = cd[['Category 3', 'Category 4', 'Category 22', 'Category 24']].mean(axis=1).round(2)
cd['Culture_Rating'] = cd[['Category 1', 'Category 6', 'Category 23']].mean(axis=1).round(2)
cd['Food_Rating'] = cd[['Category 9', 'Category 12', 'Category 21']].mean(axis=1).round(2)
cd['Nightlife_Rating'] = cd[['Category 10', 'Category 16']].mean(axis=1).round(2)

cd.head()
```

```python
cd = cd.apply(lambda col: col.str.replace('\t', '').str.strip() if col.dtype == "object" else col)
cd = cd.astype(float)
print(cd.dtypes)
cd.to_csv("clean_data_with_new_features.csv", index=False)
```

```python
new_features = ['Leisure_Rating', 'Culture_Rating', 'Food_Rating', 'Nightlife_Rating']

plt.figure(figsize=(6, 5))
sns.heatmap(cd[new_features].corr(), annot=True, fmt=".2f", cmap='coolwarm', square=True)
plt.title("Correlation Matrix of New Features")
plt.tight_layout()
plt.show()
```

```python
cd[new_features].hist(figsize=(10, 6), bins=10, color='#4C72B0', edgecolor='black')
plt.suptitle("Distribution of New Features", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
colors = ['blue', 'green', 'red', 'purple']
avg = cd[new_features].mean()
plt.figure(figsize=(9, 5))
bars = sns.barplot(x=avg.index, y=avg.values)
for i, bar in enumerate(bars.patches):
    bar.set_facecolor(colors[i])
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 0.05, f'{height:.2f}',
             ha='center', va='bottom', fontsize=10, fontweight='bold')
plt.title("Average Ratings Including Tourist Attractiveness", fontsize=14)
plt.ylabel("Average Rating")
plt.ylim(0, 5.5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
