```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("google_review_ratings.csv")
print(df.head())
```

```python
print(df.info())
```

```python
missing_values = df.isnull().sum()
print(missing_values)
print(df.columns)
#see unnamed column with all missing values. I guess this is because of the extra coma at the end of each line.
```

```python
#Delete this extra column and check again
df = df.drop(columns=df.columns[-1])
missing_values = df.isnull().sum()
print(missing_values)
print(df.columns)
```

```python
df = df.fillna(df.mean(numeric_only=True))
print("\nMissing values after filling only numeric:")
print(df.isnull().sum())

df = df.drop_duplicates()
df.drop("User", axis=1, inplace=True) #Delete "User" column

#Clear all lines in the data from unnecessary whitespace
df = df.apply(lambda col: col.str.replace('\t', '').str.strip() if col.dtype == "object" else col)
df = df.astype(float)
print(df.dtypes)
df.to_csv("cleaned_google_review_ratings.csv", index=False)
```

```python
print(df.describe())
```

```python
mean = df.mean()
median = df.median()
mode = df.mode()
std_dev = df.std()

print("Basic Statistics for Numerical Features:")
print("="*50)
for column in df:
    print(f"\nColumn: {column}")
    print(f"Mean: {mean[column]}")
    print(f"Median: {median[column]}")
    print(f"Mode: {mode[column]}")
    print(f"Standard Deviation: {std_dev[column]}")
```

```python
plt.figure(figsize=(16, 12))
df.hist(bins=30, figsize=(16, 12))
plt.subplots_adjust(hspace=0.5, wspace=0.5)
plt.show()
```

```python
plt.figure(figsize=(16, 12))
sns.boxplot(data=df, orient='h', palette='Set2')
plt.title('Box Plots')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df['Category 1'], y=df['Category 2'])
plt.title('Scatter Plot: Average ratings on churches vs resorts')
plt.xlabel('Average ratings on churches')
plt.ylabel('Average ratings on resorts')
plt.show()
#I think my dataset is not suitable for a scatter plot
```

```python
correlation_matrix = df.corr()
plt.figure(figsize=(14, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', square=True, cbar_kws={"shrink": 0.8})
plt.title('Correlation Heatmap')
plt.show()
```
